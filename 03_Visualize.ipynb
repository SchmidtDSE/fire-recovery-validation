```python
#conda activate burnseverity
import pandas as pd
import geopandas as gpd
import requests
import rasterio as rio
from matplotlib import pyplot as plt
from rasterio.plot import show_hist
import numpy as np
from rasterio.warp import calculate_default_transform, reproject, Resampling

from src import config
from src import visualization as viz
```

### Get fire boundary from Calfire

```python
fires = pd.read_csv(config.PATH_JOBS_LOG)
calfire = gpd.read_file(config.PATH_FIRES_VALIDATION)
fire_name = 'ELK TRAIL'
fire_days = 5
fire_polygon = calfire[calfire['FIRE_NAME'] == fire_name].to_crs(epsg=4326)
```

### Plot all rasters

```python
for sensor in config.SENSORS:
    for fire_days in config.POST_FIRE_PERIOD:
        try:
            viz.plot_fire(fire_name, fire_days, fire_polygon, fires, sensor=sensor)
        except ValueError as e:
            print(f'Skipping {sensor}, {fire_days} days: {e}')
```
