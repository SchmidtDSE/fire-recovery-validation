## Retrieve COGs

This notebook provides the code to retrieve COGs for individual fires for further analysis. 

```python
import pandas as pd
import geopandas as gpd
import rasterio as rio

from src import config
from src import process
```

```python
fires = pd.read_csv(config.PATH_JOBS_LOG)
calfire = gpd.read_file(config.PATH_FIRES_VALIDATION)
fire_name = 'ELK TRAIL'
fire_days = 5
sensor = 'sentinel-2'
```

```python
fire_event_name, job_id = process.lookup_job(fire_name, fire_days, 'alarm', sensor, fires)

dnbr_url = process.get_url_raster(fire_event_name, job_id, metric = 'dnbr')
rbr_url = process.get_url_raster(fire_event_name, job_id, metric = 'rbr')

dnbr_reproj, dnbr_extent = process.get_raster_as_lonlat(dnbr_url)
rbr_reproj, rbr_extent = process.get_raster_as_lonlat(rbr_url)
dnbr_reproj

import os
os.makedirs('output', exist_ok=True)
```

```python
dnbr_reproj.rio.to_raster(f'output/{fire_name}_{fire_days}_days_{sensor}_dnbr.tif')
rbr_reproj.rio.to_raster(f'output/{fire_name}_{fire_days}_days_{sensor}_rbr.tif')
```
